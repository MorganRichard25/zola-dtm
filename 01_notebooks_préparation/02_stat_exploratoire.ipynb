## Synthèse

Le corpus est composé de **5 327 paquets**. Un paquet contient en moyenne **785 mots**, avec une médiane de **753 mots** et un écart-type de **212 mots**.

Les valeurs extrêmes sont rares :

- 12 paquets contiennent moins de 300 mots, soit environ 0,23 % du corpus ;
- 24 paquets contiennent plus de 1 500 mots, soit environ 0,45 % du corpus ;
- 1 paquet contient plus de 2 000 mots, soit environ 0,02 % du corpus.

Ces observations ne justifient pas d'exclure des paquets à ce stade. Les valeurs extrêmes sont conservées, tout en restant à surveiller lors de la vectorisation et de la modélisation.
